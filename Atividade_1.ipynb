# Parte A

Funções utlizadas na parte A

```python
# Define uma função para truncar um número em determinada quantidade de casas.
def truncar(numero, casas):
    # Calcula o fator que desloca a vírgula decimal.
    fator = 10 ** casas

    # Trunca o número deslocado e devolve a vírgula à posição original.
    return math.trunc(numero * fator) / fator
```

Listas Utilizadas na parte A

```python
n = [0, 1, 2, 3, 4]

x = [3.14159265, 98.76543, 0.00498765]
arredondados = []
truncados = []

erros_absoluto_round = []
erros_relativo_round = []
erros_percentual_round = []

erros_absoluto_trunc = []
erros_relativo_trunc = []
erros_percentual_trunc = []
```

```python
for num in x:
  texto = f"Aproximação por truncamento: {num}"
  largura = 80
  print("\n")
  print("=" * largura)
  print(texto.center(largura))
  print("=" * largura)


  for i in n:
    truncado = truncar(num, i)
    truncados.append(truncado)
    print(f'O número {num} truncado em {i} casas é igual a {truncado}')
```

Portanto, para a regra convencional de arredondamento ao valor mais próximo, o erro absoluto do arredondamento é menor ou igual ao erro do truncamento.

## B.1 Representação das medidas

### Séries parte B

```python
serie_original = np.array([121.7, 120.4, 122.1, 119.8, 121.2, 120.9, 122.4, 121.0, 120.2, 121.5])
serie_round = np.array([round(m) for m in serie_original])
serie_trunc = np.array([int(truncar(m, 0)) for m in serie_original])
```

### Cálculo da média, mínimo, máximo, amplitude e desvio-padrão amostral

```python
# Cálculo para os dados originais
orginal_media = serie_original.mean()
original_max = serie_original.max()
original_min = serie_original.min()
amplitude_original = original_max - original_min
desvio_original = np.std(serie_original, ddof=1)


# Cálculo para os dados arredondados
round_media = serie_round.mean()
round_max = serie_round.max()
round_min = serie_round.min()
round_amplitude = round_max - round_min
round_desvio = np.std(serie_round, ddof=1)


# Cálculo para os dados truncados
trunc_media = serie_trunc.mean()
trunc_max = serie_trunc.max()
trunc_min = serie_trunc.min()
trunc_amplitude = trunc_max - trunc_min
trunc_desvio = np.std(serie_trunc, ddof=1)
```

### Erros Média Arredondamento

```python
# Erro Absoluto entre Média com dados original e arredondados
erro_absoluto_round = abs(orginal_media - round_media)

# Erro Relativo entre Média com dados original e arredondados
erro_relativo_round = erro_absoluto_round / abs(orginal_media)

# Erro Percentual entre Média com dados original e arredondados
erro_percentual_round = 100.0 * erro_relativo
```

### Erros Média Truncamento

```python
# Erro Absoluto entre Média com dados original e truncamento
erro_absoluto_trunc = abs(orginal_media - trunc_media)

# Erro Relativo entre Média com dados original e truncamento
erro_relativo_trunc = erro_absoluto_trunc / abs(orginal_media)

# Erro Percentual entre Média com dados original e truncamento
erro_percentual_trunc = 100.0 * erro_relativo
```

Entretanto, mesmo o arredondamento alterou o desvio-padrão e a amplitude dos dados. Portanto, para preservar melhor as características estatísticas da amostra, é preferível manter as casas decimais originais ou utilizar uma representação com maior precisão.

## C.1 Precisão das variáveis

### Função do Cálculo da Vazão

```python
def calcular_vazao(r, delta_p, mu, L):
    r = r / 1000
    Q = (np.pi * r**4 * delta_p) / (8 * mu * L)
    return Q
```

```python
r = 0.80
L = 0.20
mu = 0.0035
delta_p = 1200
```

### Cálculo de Q com os dados fornecidos

```python
Q = calcular_vazao(r, delta_p, mu, L)
Q
```

### Calculo de Q usando r arredondado e truncado para uma casa decimal em mm

```python
r_round = round(r,1)
r_trunc = truncar(r, 1)

Q_r_round = calcular_vazao(r_round, delta_p, mu, L)
Q_r_trunc = calcular_vazao(r_trunc, delta_p, mu, L)

texto = f"Calculo de Q "
largura = 70
print("=" * largura)
print(texto.center(largura))
print("=" * largura)

print(f'Q = {Q_r_round}, para arredondamento de r = {r_round}')
print(f'Q = {Q_r_trunc}, para truncamento de r = {r_trunc}')


# Erro Absoluto
Ea_Q_r_round = abs(Q - Q_r_round)
Ea_Q_r_trunc = abs(Q - Q_r_trunc)

texto = f"Erro Absoluto"
largura = 70
print("\n")
print("=" * largura)
print(texto.center(largura))
print("=" * largura)
print(f'Erro Absoluto entre o valor original e arredondado de r: {Ea_Q_r_round}')
print(f'Erro Absoluto entre o valor original e truncado de r: {Ea_Q_r_trunc}')


# Erro Relativo
Er_Q_r_round = Ea_Q_r_round / abs(Q)
Er_Q_r_trunc = Ea_Q_r_trunc / abs(Q)


texto = f"Erro Relativo"
largura = 70
print("\n")
print("=" * largura)
print(texto.center(largura))
print("=" * largura)
print(f'Erro Relativo entre o valor original e arredondado de r: {Er_Q_r_round}')
print(f'Erro Relativo entre o valor original e truncado de r: {Er_Q_r_trunc}')


# Erro Percentual
Ep_Q_r_round = 100.0 * Er_Q_r_round
Ep_Q_r_trunc = 100.0 * Er_Q_r_trunc

texto = f"Erro Percentual"
largura = 70
print("\n")
print("=" * largura)
print(texto.center(largura))
print("=" * largura)
print(f'Erro Percentual entre o valor original e arredondado de r: {Ep_Q_r_round}')
print(f'Erro Percentual entre o valor original e truncado de r: {Ep_Q_r_trunc}')
```

```python
# Cálculo de Q

for r in raios:

    # Q usando o raio original
    Q = calcular_vazao(r, delta_p, mu, L)
    Q_referencia.append(Q)

    # Raio arredondado para uma casa decimal
    r_round = round(r, 1)

    # Q usando o raio arredondado
    Q_round = calcular_vazao(r_round, delta_p, mu, L)
    Q_arredondado.append(Q_round)

    # Raio truncado para uma casa decimal
    r_trunc = truncar(r, 1)

    # Q usando o raio truncado
    Q_trunc = calcular_vazao(r_trunc, delta_p, mu, L)
    Q_truncado.append(Q_trunc)



# CONVERTER LISTAS PARA ARRAYS NUMPY

Q_referencia = np.array(Q_referencia)
Q_arredondado = np.array(Q_arredondado)
Q_truncado = np.array(Q_truncado)



# 1. Q EM FUNÇÃO DE r

plt.figure(figsize=(10, 6))

plt.plot(
    raios,
    Q_referencia,
    marker='o',
    label='Q referência'
)

plt.plot(
    raios,
    Q_arredondado,
    marker='s',
    linestyle='--',
    label='Q com raio arredondado'
)

plt.plot(
    raios,
    Q_truncado,
    marker='^',
    linestyle=':',
    label='Q com raio truncado'
)

plt.xlabel('Raio r (mm)')
plt.ylabel('Vazão Q')
plt.title('Q em função de r')

plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()



# Erro absoluto
erro_absoluto_round = np.abs(
    Q_referencia - Q_arredondado
)

erro_absoluto_trunc = np.abs(
    Q_referencia - Q_truncado
)


# Erro relativo
erro_relativo_round = (
    erro_absoluto_round / np.abs(Q_referencia)
)

erro_relativo_trunc = (
    erro_absoluto_trunc / np.abs(Q_referencia)
)


# Erro percentual
erro_percentual_round = (
    erro_relativo_round * 100
)

erro_percentual_trunc = (
    erro_relativo_trunc * 100
)




# 2. ERRO ABSOLUTO DE Q EM FUNÇÃO DE r

plt.figure(figsize=(10, 6))

plt.plot(
    raios,
    erro_absoluto_round,
    marker='o',
    label='Arredondamento'
)

plt.plot(
    raios,
    erro_absoluto_trunc,
    marker='s',
    label='Truncamento'
)

plt.xlabel('Raio r (mm)')
plt.ylabel('Erro absoluto de Q')
plt.title('Erro absoluto de Q em função de r')

plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()




# 3. ERRO RELATIVO DE Q EM FUNÇÃO DE r

plt.figure(figsize=(10, 6))

plt.plot(
    raios,
    erro_relativo_round,
    marker='o',
    label='Arredondamento'
)

plt.plot(
    raios,
    erro_relativo_trunc,
    marker='s',
    label='Truncamento'
)

plt.xlabel('Raio r (mm)')
plt.ylabel('Erro relativo de Q')
plt.title('Erro relativo de Q em função de r')

plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()



# 4. ERRO PERCENTUAL DE Q EM FUNÇÃO DE r

plt.figure(figsize=(10, 6))

plt.plot(
    raios,
    erro_percentual_round,
    marker='o',
    label='Arredondamento'
)

plt.plot(
    raios,
    erro_percentual_trunc,
    marker='s',
    label='Truncamento'
)

plt.xlabel('Raio r (mm)')
plt.ylabel('Erro percentual de Q (%)')
plt.title('Erro percentual de Q em função de r')

plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```

# Parte D – escoamento em um nanocateter

```python
# Dados
r = 50e-9              # raio em metros
L = 10e-6              # comprimento em metros
mu = 3.5e-3            # viscosidade em Pa.s
delta_p = 5000         # diferença de pressão em Pa


# Q DE REFERÊNCIA
Q_referencia = calcular_vazao(
    r,
    delta_p,
    mu,
    L
)

print(f"Q de referência = {Q_referencia:.6e} m³/s")



# VARIAÇÃO DO ERRO NO RAIO

# Delta r de 0,1 até 5,0 nm
delta_r_nm = np.arange(0.1, 5.1, 0.1)

# Conversão para metros
delta_r = delta_r_nm * 1e-9


# r- E r+

r_menos = r - delta_r
r_mais = r + delta_r



# VAZÕES

Q_menos = calcular_vazao(
    r_menos,
    delta_p,
    mu,
    L
)

Q_mais = calcular_vazao(
    r_mais,
    delta_p,
    mu,
    L
)



# ERROS ABSOLUTOS

erro_absoluto_menos = np.abs(
    Q_menos - Q_referencia
)

erro_absoluto_mais = np.abs(
    Q_mais - Q_referencia
)


# Erro absoluto máximo
erro_absoluto_max = np.maximum(
    erro_absoluto_menos,
    erro_absoluto_mais
)


# ERROS RELATIVOS

erro_relativo_menos = (
    erro_absoluto_menos /
    np.abs(Q_referencia)
)

erro_relativo_mais = (
    erro_absoluto_mais /
    np.abs(Q_referencia)
)


# Erro relativo máximo
erro_relativo_max = np.maximum(
    erro_relativo_menos,
    erro_relativo_mais
)


# ERROS PERCENTUAIS

erro_percentual_max = (
    erro_relativo_max * 100
)


# APROXIMAÇÃO LINEAR

# ΔQ/Q ≈ 4 Δr/r

erro_relativo_linear = (
    4 * delta_r / r
)

erro_percentual_linear = (
    erro_relativo_linear * 100
)



# 1. GRÁFICO Q(r-), Q(r) e Q(r+)

plt.figure(figsize=(10, 6))

plt.plot(
    delta_r_nm,
    Q_menos,
    label='Q(r-)'
)

plt.plot(
    delta_r_nm,
    np.full(len(delta_r_nm), Q_referencia),
    label='Q(r)'
)

plt.plot(
    delta_r_nm,
    Q_mais,
    label='Q(r+)'
)

plt.xlabel('Δr (nm)')
plt.ylabel('Q (m³/s)')
plt.title('Q(r−), Q(r) e Q(r+) em função de Δr')

plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()


# 2. GRÁFICO DO ERRO ABSOLUTO MÁXIMO

plt.figure(figsize=(10, 6))

plt.plot(
    delta_r_nm,
    erro_absoluto_max,
    marker='o'
)

plt.xlabel('Δr (nm)')
plt.ylabel('Erro absoluto máximo de Q (m³/s)')
plt.title('Erro absoluto máximo de Q em função de Δr')

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()



# 3. GRÁFICO DO ERRO RELATIVO MÁXIMO

plt.figure(figsize=(10, 6))

plt.plot(
    delta_r_nm,
    erro_relativo_max,
    marker='o',
    label='Erro relativo máximo'
)

plt.plot(
    delta_r_nm,
    erro_relativo_linear,
    linestyle='--',
    label='Aproximação linear: 4Δr/r'
)

plt.xlabel('Δr (nm)')
plt.ylabel('Erro relativo')
plt.title('Erro relativo máximo de Q em função de Δr')

plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()



# 4. GRÁFICO DO ERRO PERCENTUAL MÁXIMO

plt.figure(figsize=(10, 6))

plt.plot(
    delta_r_nm,
    erro_percentual_max,
    marker='o',
    label='Erro percentual máximo'
)

plt.plot(
    delta_r_nm,
    erro_percentual_linear,
    linestyle='--',
    label='Aproximação linear: 4Δr/r'
)

plt.xlabel('Δr (nm)')
plt.ylabel('Erro percentual máximo (%)')
plt.title('Erro percentual máximo de Q em função de Δr')

plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```

## D.2 Precisão computacional

```python
# float32
r32 = np.float32(r)
L32 = np.float32(L)
mu32 = np.float32(mu)
delta_p32 = np.float32(delta_p)

r4_32 = r32**4

Q32 = calcular_vazao(
    r32,
    delta_p32,
    mu32,
    L32
)


# float64
r64 = np.float64(r)
L64 = np.float64(L)
mu64 = np.float64(mu)
delta_p64 = np.float64(delta_p)

r4_64 = r64**4

Q64 = calcular_vazao(
    r64,
    delta_p64,
    mu64,
    L64
)



# DIFERENÇAS

erro_absoluto = abs(Q64 - Q32)

erro_relativo = erro_absoluto / abs(Q64)


# EPSILON DE CADA PRECISÃO

eps32 = np.finfo(np.float32).eps
eps64 = np.finfo(np.float64).eps


# RESULTADOS

print("FLOAT32")
print(f"r     = {r32:.15e}")
print(f"r^4   = {r4_32:.15e}")
print(f"Q     = {Q32:.15e}")
print(f"eps   = {eps32:.15e}")


print("\nFLOAT64")
print(f"r     = {r64:.15e}")
print(f"r^4   = {r4_64:.15e}")
print(f"Q     = {Q64:.15e}")
print(f"eps   = {eps64:.15e}")


print("\nCOMPARAÇÃO")
print(f"Erro absoluto = {erro_absoluto:.15e}")
print(f"Erro relativo = {erro_relativo:.15e}")
print(f"Erro percentual = {erro_relativo * 100:.15e}%")
```

## D.3 Fenômeno físico

A redução da escala do escoamento, como no caso de um nanocateter, aumenta a razão entre a superfície e o volume, fazendo com que os efeitos de viscosidade e de interação com as paredes tenham maior influência sobre o escoamento. Assim, fenômenos que podem ser desprezados em escalas maiores podem se tornar relevantes em dimensões muito pequenas, podendo limitar a validade de modelos simplificados, como o de Hagen–Poiseuille. Dessa forma, um resultado apresentado com muitas casas decimais pode ser numericamente preciso, no sentido de ter sido calculado com alta precisão aritmética, mas ainda ser fisicamente inadequado caso o modelo utilizado não represente corretamente os fenômenos envolvidos nessa escala. Portanto, precisão aritmética e validade do modelo são conceitos diferentes: aumentar a precisão dos cálculos não garante que o modelo físico utilizado seja adequado ao problema.
